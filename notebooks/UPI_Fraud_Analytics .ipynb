```python
# DATA CLEANING — Step 6: Handle time_since_last_txn_min
# determine how to treat the 11,212 NULL values by comparing fraud rates for available vs. missing values.

from pyspark.sql import functions as F

transactions_df = spark.table(
    "UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS"
)

transactions_clean_df = (
    transactions_df
    .withColumn(
        "time_since_last_txn_min",
        F.when(
            F.col("time_since_last_txn_min") < 0,
            None
        ).otherwise(F.col("time_since_last_txn_min"))
    )
    .withColumn(
        "transaction_velocity",
        F.coalesce(
            F.col("transaction_velocity"),
            F.lit(0)
        )
    )
)

median_amount_deviation = transactions_clean_df.select(
    F.expr("percentile_approx(amount_deviation_score, 0.5)")
).first()[0]

transactions_clean_df = transactions_clean_df.withColumn(
    "amount_deviation_score",
    F.coalesce(
        F.col("amount_deviation_score"),
        F.lit(median_amount_deviation)
    )
)

display(
    transactions_clean_df
    .withColumn(
        "time_data_status",
        F.when(
            F.col("time_since_last_txn_min").isNull(),
            "Missing"
        ).otherwise("Available")
    )
    .groupBy("time_data_status")
    .agg(
        F.count("*").alias("transaction_count"),
        F.sum("is_fraud").alias("fraud_count"),
        F.round(
            F.avg("is_fraud") * 100, 2
        ).alias("fraud_rate_percent")
    )
    .orderBy("time_data_status")
)
```

```python

# # Diagnostic Analytics — Fraud Risk Indicator Analysis
# Identify whether the key fraud risk indicators are associated with higher fraud. We’ll start with New Device vs Fraud

from pyspark.sql import functions as F

transactions_df = spark.table("UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS")
new_device_analysis = transactions_df.groupBy("new_device_flag") \
    .agg(
        F.count("*").alias("Transaction_Count"),
        F.sum("is_fraud").alias("Fraud_Count"),
        F.round(F.avg("is_fraud") * 100, 2).alias("Fraud_Rate_Percent")
    ) \
    .orderBy("new_device_flag")

display(new_device_analysis)

```

```python
# Diagnostic Analytics — Step 2 -Analyze IP Location Mismatch vs Fraud to see whether transactions with a location mismatch have a higher fraud rate

from pyspark.sql import functions as F

transactions_df = spark.table("UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS")

ip_mismatch_analysis = transactions_df.groupBy("ip_location_mismatch") \
    .agg(
        F.count("*").alias("Transaction_Count"),
        F.sum("is_fraud").alias("Fraud_Count"),
        F.round(F.avg("is_fraud") * 100, 2).alias("Fraud_Rate_Percent")
    ) \
    .orderBy("ip_location_mismatch")

display(ip_mismatch_analysis)

```

```python
# Diagnostic Analytics — Step 3-Analyze Failed Attempts vs Fraud to determine whether repeated failed attempts are associated with higher fraud risk.

from pyspark.sql import functions as F

transactions_df = spark.table("UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS")

failed_attempts_analysis = transactions_df.groupBy("failed_attempts_last_24h") \
    .agg(
        F.count("*").alias("Transaction_Count"),
        F.sum("is_fraud").alias("Fraud_Count"),
        F.round(F.avg("is_fraud") * 100, 2).alias("Fraud_Rate_Percent")
    ) \
    .orderBy("failed_attempts_last_24h")

display(failed_attempts_analysis)
```

```python
# Diagnostic Analytics — Step 4-Analyze Transaction Velocity vs Fraud to determine whether higher transaction activity is associated with increased fraud risk.

from pyspark.sql import functions as F

transactions_df = spark.table("UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS")

velocity_analysis = transactions_df.groupBy("transaction_velocity") \
    .agg(
        F.count("*").alias("Transaction_Count"),
        F.sum("is_fraud").alias("Fraud_Count"),
        F.round(F.avg("is_fraud") * 100, 2).alias("Fraud_Rate_Percent")
    ) \
    .orderBy("transaction_velocity")

display(velocity_analysis)
```

```python
# Diagnostic Analytics — Step 5 -Analyze Amount Deviation Score vs Fraud using the actual amount_deviation_score(fraud risk indicator) field to determine whether transactions with unusually high amounts are associated with higher fraud risk.

from pyspark.sql import functions as F

transactions_df = spark.table("UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS")
amount_deviation_analysis = transactions_df.groupBy(
    F.when(F.col("amount_deviation_score") < 0.5, "Low")
     .when(F.col("amount_deviation_score") < 1, "Medium")
     .otherwise("High")
     .alias("Deviation_Category")
).agg(
    F.count("*").alias("Transaction_Count"),
    F.sum("is_fraud").alias("Fraud_Count"),
    F.round(F.avg("is_fraud") * 100, 2).alias("Fraud_Rate_Percent")
)

display(amount_deviation_analysis)

```

```python
# Diagnostic Analytics — Step 6 -Complete the diagnostic analysis by checking fraud patterns over time of day. This supports the “Fraud Transactions by Hour of Day” chart in Dashboard 2.

from pyspark.sql import functions as F

transactions_df = spark.table("UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS")
hourly_fraud = transactions_df.groupBy("hour_of_day") \
    .agg(
        F.count("*").alias("Transaction_Count"),
        F.sum("is_fraud").alias("Fraud_Count"),
        F.round(F.avg("is_fraud") * 100, 2).alias("Fraud_Rate_Percent")
    ) \
    .orderBy("hour_of_day")

display(hourly_fraud)
```

```python
# DASHBOARD 2--------Fraud & Risk Analysis: KPI Calculations (All KPI's)

from pyspark.sql import functions as F

transactions_df = spark.table("UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS")
users_df = spark.table("UPI_SNOWFLAKE_CATALOG.RAW_DATA.USERS")

fraud_kpis = transactions_df.select(
    F.sum("is_fraud").alias("Fraudulent_Transactions"),
    F.round(F.avg("is_fraud") * 100, 2).alias("Fraud_Rate_Percent"),
    F.sum(F.when(F.col("is_fraud") == 1, F.col("amount")).otherwise(0)).alias("Fraud_Transaction_Value"),
    F.sum(F.when(F.col("new_device_flag") == 1, 1).otherwise(0)).alias("New_Device_Transactions")
)

high_risk_users = users_df.filter(
    F.col("is_high_risk_user") == 1
).agg(
    F.countDistinct("user_id").alias("High_Risk_Users")
)

fraud_kpis.crossJoin(high_risk_users).show()

```

```python
# 2 — Step 2: Fraud Trend (CHART) -gives us the daily fraud transaction count and fraud value needed for the Dashboard 2 trend chart.

from pyspark.sql import functions as F

transactions_df = spark.table("UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS")

fraud_trend = transactions_df.filter(F.col("is_fraud") == 1) \
    .groupBy("date") \
    .agg(
        F.count("*").alias("Fraudulent_Transactions"),
        F.sum("amount").alias("Fraud_Transaction_Value")
    ) \
    .orderBy("date")

display(fraud_trend)
```

```python
# Dashboard 2 — Step 3: Fraud Rate by Payment App (chart)-Compare the fraud rate across payment app
#gives us:Transaction Count, Fraud Count ,Fraud Rate (%)

from pyspark.sql import functions as F

transactions_df = spark.table("UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS")

fraud_app_analysis = transactions_df.groupBy("payment_app") \
    .agg(
        F.count("*").alias("Transaction_Count"),
        F.sum("is_fraud").alias("Fraud_Count"),
        F.round(F.avg("is_fraud") * 100, 2).alias("Fraud_Rate_Percent")
    ) \
    .orderBy(F.desc("Fraud_Rate_Percent"))

display(fraud_app_analysis)
```

```python
# Dashboard 2 — Step 4: Fraud by Device Type(compare fraudulent transactions across device types.)

from pyspark.sql import functions as F

transactions_df = spark.table("UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS")

fraud_device_analysis = transactions_df.groupBy("device_type") \
    .agg(
        F.sum("is_fraud").alias("Fraud_Count"),
        F.round(F.avg("is_fraud") * 100, 2).alias("Fraud_Rate_Percent")
    ) \
    .orderBy(F.desc("Fraud_Count"))

display(fraud_device_analysis)
```

```python
# Dashboard 2 — Step 5: Fraud Transactions by Hour of Day -analyzing when fraudulent transactions occur during the day.

from pyspark.sql import functions as F

transactions_df = spark.table("UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS")

hourly_fraud = transactions_df.filter(F.col("is_fraud") == 1) \
    .groupBy("hour_of_day") \
    .agg(
        F.count("*").alias("Fraud_Count"),
        F.sum("amount").alias("Fraud_Transaction_Value")
    ) \
    .orderBy("hour_of_day")

display(hourly_fraud)
```

```python
#Predictive Analytics — Step 1

# Objective -predict whether a UPI transaction is potentially fraudulent based on transaction behaviour, amount patterns, time characteristics, and key risk indicators.

#Purpose: Prepare the dataset for fraud prediction. Our target variable is is_fraud, and we’ll use the transaction/risk fields already present in the dataset.


from pyspark.sql import functions as F

transactions_df = spark.table("UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS")

model_df = transactions_df.select(
    "amount",
    "hour_of_day",
    "is_weekend",
    "is_night_transaction",
    "new_device_flag",
    "ip_location_mismatch",
    "failed_attempts_last_24h",
    "transaction_velocity",
    "amount_deviation_score",
    "is_fraud"
).dropna()

print("Rows available for modeling:", model_df.count())
display(model_df.limit(10))
```

```python
# Predictive Analytics — Step 2
# Purpose: Split the prepared data into training and testing datasets so the model can learn from one portion and be evaluated on unseen data.

train_df, test_df = model_df.randomSplit([0.8, 0.2], seed=42)

print("Training rows:", train_df.count())
print("Testing rows:", test_df.count())
```

```python
#Predictive Analytics — Step 3
#Purpose: Convert the selected input columns into a single feature vector that Spark ML can use for fraud classification.

from pyspark.ml.feature import VectorAssembler

features = [
    "amount",
    "hour_of_day",
    "is_weekend",
    "is_night_transaction",
    "new_device_flag",
    "ip_location_mismatch",
    "failed_attempts_last_24h",
    "transaction_velocity",
    "amount_deviation_score"
]

assembler = VectorAssembler(
    inputCols=features,
    outputCol="features"
)

train_data = assembler.transform(train_df)
test_data = assembler.transform(test_df)

print("Feature vector created successfully")
```

```python
#Predictive Analytics — Step 4

#Purpose: Train the fraud classification model using the training data.
# We’ll use Random Forest, which is suitable for this classification problem and can capture non-linear relationships between the risk indicators and fraud.

from pyspark.ml.classification import RandomForestClassifier

rf = RandomForestClassifier(
    featuresCol="features",
    labelCol="is_fraud",
    numTrees=100,
    seed=42
)

model = rf.fit(train_data)

print("Random Forest model trained successfully")
```

```python
# Predictive Analytics — Step 5
#Prepare the modeling data, split it, create features, train Random Forest, and generate predictions in one cell.

from pyspark.sql import functions as F
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.classification import RandomForestClassifier

transactions_df = spark.table(
    "UPI_SNOWFLAKE_CATALOG.RAW_DATA.TRANSACTIONS"
)

model_df = transactions_df.select(
    "amount",
    "hour_of_day",
    "is_weekend",
    "is_night_transaction",
    "new_device_flag",
    "ip_location_mismatch",
    "failed_attempts_last_24h",
    "transaction_velocity",
    "amount_deviation_score",
    "is_fraud"
).dropna()

train_df, test_df = model_df.randomSplit([0.8, 0.2], seed=42)

features = [
    "amount",
    "hour_of_day",
    "is_weekend",
    "is_night_transaction",
    "new_device_flag",
    "ip_location_mismatch",
    "failed_attempts_last_24h",
    "transaction_velocity",
    "amount_deviation_score"
]

assembler = VectorAssembler(
    inputCols=features,
    outputCol="features"
)

train_data = assembler.transform(train_df)
test_data = assembler.transform(test_df)

rf = RandomForestClassifier(
    featuresCol="features",
    labelCol="is_fraud",
    numTrees=100,
    seed=42
)

model = rf.fit(train_data)

predictions = model.transform(test_data)

display(
    predictions.select(
        "is_fraud",
        "prediction",
        "probability"
    ).limit(20)
)

```

```python
#Predictive Analytics — Step 6
#Purpose: Evaluate the fraud prediction model using the key metrics planned for the project: Precision, Recall, F1-Score, and ROC-AUC.

from pyspark.ml.evaluation import MulticlassClassificationEvaluator, BinaryClassificationEvaluator

precision = MulticlassClassificationEvaluator(
    labelCol="is_fraud",
    predictionCol="prediction",
    metricName="precisionByLabel"
).setMetricLabel(1.0).evaluate(predictions)

recall = MulticlassClassificationEvaluator(
    labelCol="is_fraud",
    predictionCol="prediction",
    metricName="recallByLabel"
).setMetricLabel(1.0).evaluate(predictions)

f1 = MulticlassClassificationEvaluator(
    labelCol="is_fraud",
    predictionCol="prediction",
    metricName="f1"
).evaluate(predictions)

roc_auc = BinaryClassificationEvaluator(
    labelCol="is_fraud",
    rawPredictionCol="rawPrediction",
    metricName="areaUnderROC"
).evaluate(predictions)

print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-Score:  {f1:.4f}")
print(f"ROC-AUC:   {roc_auc:.4f}")
```

```python
# Prescriptive Analytics — Step 1
#Purpose: Extract the fraud probability from the model output and assign Low / Medium / High risk.

#Low Risk → Normal processing
#Medium Risk → Additional verification
#High Risk → Flag for investigation

from pyspark.sql import functions as F
from pyspark.ml.functions import vector_to_array

risk_scored = predictions.withColumn(
    "fraud_probability",
    vector_to_array("probability")[1]
).withColumn(
    "risk_level",
    F.when(F.col("fraud_probability") < 0.30, "Low")
     .when(F.col("fraud_probability") < 0.70, "Medium")
     .otherwise("High")
)

display(
    risk_scored.select(
        "is_fraud",
        "prediction",
        "fraud_probability",
        "risk_level"
    ).limit(20)
)


#Note: The 0.30 and 0.70 cutoffs are business-policy thresholds for this project,
```

```python
#Prescriptive Analytics — Step 2
#Purpose: Summarize the transactions into the three risk-action categories so we can quantify the recommended business actions.

risk_summary = risk_scored.groupBy("risk_level") \
    .agg(
        F.count("*").alias("Transaction_Count"),
        F.sum("is_fraud").alias("Fraud_Count"),
        F.round(F.avg("is_fraud") * 100, 2).alias("Fraud_Rate_Percent")
    ) \
    .orderBy("risk_level")

display(risk_summary)
```
